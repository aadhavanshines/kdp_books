## Arrays and Vectorized Operations

`arrays.py`

```python
import numpy as np

prices = np.array([45.0, 60.0, 24.0, 70.0, 35.0])
quantities = np.array([2, 1, 4, 1, 3])

print(prices.dtype, prices.shape)
print(prices * quantities)  # element by element
print((prices * quantities).sum())  # total
print(prices[prices > 40])  # a boolean mask selects items
print(np.round(prices * 1.08, 2))  # every price plus 8% tax

grid = np.arange(12).reshape(3, 4)
print(grid)
print(grid.sum(axis=0))  # down each column
print(grid.sum(axis=1))  # across each row
```

## Broadcasting

`broadcasting.py`

```python
import numpy as np

# Monthly sales (rows) for three stores (columns)
sales = np.array([[120, 95, 60], [130, 90, 75], [150, 110, 80]])
targets = np.array([125, 100, 70])  # one target per store

print(sales - targets)  # the 1-D targets stretch to every row
print(sales >= targets)
print((sales >= targets).mean(axis=0).round(2))  # months on target
print((sales / sales.sum(axis=1, keepdims=True)).round(2))  # shares
```

## Random Numbers and Simple Statistics

`random_stats.py`

```python
import numpy as np

rng = np.random.default_rng(42)  # a fixed seed: the same numbers each run
heights = rng.normal(loc=170, scale=8, size=10_000)

print(f"mean   {heights.mean():.1f}")
print(f"std    {heights.std():.1f}")
print(f"median {np.median(heights):.1f}")
print("5th and 95th percentiles:", np.percentile(heights, [5, 95]).round(1))
print("share above 185 cm:", (heights > 185).mean())

dice = rng.integers(1, 7, size=(100_000, 2)).sum(axis=1)
counts = np.bincount(dice)[2:]
print("two dice, sums 2 to 12:")
print(counts)
print("most common sum:", counts.argmax() + 2)
```

## melt and pivot

`melt_pivot.py`

```python
import pandas as pd

wide = pd.DataFrame(
    {
        "store": ["Avon", "Bray"],
        "Jan": [120, 95],
        "Feb": [130, 90],
        "Mar": [150, 110],
    }
)
print(wide)

long = wide.melt(id_vars="store", var_name="month", value_name="sales")
print(long)

back = long.pivot(index="store", columns="month", values="sales")
print(back[["Jan", "Feb", "Mar"]])
```

## stack and unstack

`stack_unstack.py`

```python
import pandas as pd

orders = pd.read_csv("../data/orders.csv")
units = orders.groupby(["region", "channel"])["quantity"].sum()
print(units.head(4))  # a Series with a two-level index
print(units.unstack())  # inner level becomes columns
print(units.unstack().stack().equals(units))
```

## String Methods on Whole Columns

`text_methods.py`

```python
import pandas as pd

survey = pd.read_csv("../data/survey.csv", keep_default_na=False)
comments = survey["comment"]

print(comments.str.len().describe().round(1))
print()
print("mention 'slow':", comments.str.contains("slow").sum())
print(
    "mention price or value:",
    comments.str.contains(r"price|value", regex=True).sum(),
)
print()
words = comments.str.lower().str.split().explode()
stopwords = {"to", "is", "was", "too", "the"}
print(words[~words.isin(stopwords)].value_counts().head(5))
```

## Extracting and Categorizing

`extract.py`

```python
import pandas as pd

budget = pd.read_csv("../data/budget.csv")

budget["store_no"] = budget["description"].str.extract(r"#(\d+)")
rules = {
    "Groceries": r"FRESHMART",
    "Coffee": r"CAFE",
    "Transport": r"TRANSIT|RIDESHARE",
    "Housing": r"RENT|POWER",
    "Income": r"SALARY",
}
budget["category"] = "Other"
for name, pattern in rules.items():
    budget.loc[budget["description"].str.contains(pattern), "category"] = (
        name
    )

print(budget[budget["store_no"].notna()].head(3))
print(budget.groupby("category")["amount"].sum().round(2).sort_values())
```

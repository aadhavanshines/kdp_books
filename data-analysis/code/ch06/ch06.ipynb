## Columns, Labels and Positions

`selecting.py`

```python
import pandas as pd

orders = pd.read_csv("../data/orders.csv", index_col="order_id")

print(orders["product"].head(3))  # one column: a Series
print(orders[["product", "quantity"]].head(3))  # a list: a DataFrame
print(orders.loc["L00010", "product"])  # row and column labels
print(orders.loc["L00010":"L00012", ["region", "quantity"]])  # label slice
print(orders.iloc[0, :3].tolist())  # positions
```

## Filtering Rows With Conditions

`filtering.py`

```python
import pandas as pd

orders = pd.read_csv("../data/orders.csv", parse_dates=["order_date"])

big = orders[orders["quantity"] >= 5]
print("orders of 5 or more items:", len(big))

west_lamps = orders[
    (orders["region"] == "West") & (orders["category"] == "Lighting")
]
print("West lighting orders:", len(west_lamps))

promo = orders[orders["discount"].isin([0.1, 0.2])]
print("discounted orders:", len(promo))

december = orders[orders["order_date"].between("2025-12-01", "2025-12-31")]
print("December orders:", len(december))

print(
    orders.query("channel == 'app' and quantity > 4")[
        ["order_id", "product", "quantity"]
    ].head(3)
)
```

## Sorting and Top Values

`sorting.py`

```python
import pandas as pd

orders = pd.read_csv("../data/orders.csv")
orders["revenue"] = orders["quantity"] * orders["unit_price"]

cols = ["order_id", "product", "quantity", "revenue"]
# Ties at the top are broken by order_id, so the result never varies
top = orders.sort_values(["revenue", "order_id"], ascending=[False, True])
print(top[cols].head(3))
print(orders.nlargest(3, "quantity")[cols])
print(
    orders.sort_values(["region", "revenue"], ascending=[True, False])
    .groupby("region")
    .head(1)[["region"] + cols[1:]]
)
```

## Series

`series_basics.py`

```python
import pandas as pd

stock = pd.Series(
    [12, 0, 7, 25], index=["mugs", "lamps", "vases", "frames"]
)
print(stock)
print()
print(stock["vases"])  # by label
print(stock.iloc[0])  # by position
print(stock[stock > 5])  # a mask keeps matching items
print(stock.sum(), stock.mean(), stock.idxmax())
```

## DataFrames

`dataframe_basics.py`

```python
import pandas as pd

orders = pd.read_csv("../data/orders.csv", parse_dates=["order_date"])

print(orders.shape)
print(list(orders.columns[:5]))
print(list(orders.columns[5:]))
print()
print(orders.head(3)[["order_id", "order_date", "product", "quantity"]])
print()
print(orders.dtypes)
```

## A Quick Summary of Any Table

`describe.py`

```python
import pandas as pd

orders = pd.read_csv("../data/orders.csv")

print(orders[["quantity", "unit_price", "discount"]].describe().round(2))
print()
print(orders["channel"].value_counts())
print()
print(orders["category"].value_counts(normalize=True).round(3))
```

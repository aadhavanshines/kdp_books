## A Plausible, Wrong Answer

`ai_suggestion.py`

```python
import pandas as pd

orders = pd.read_csv("../data/orders.csv")
orders["revenue"] = orders["quantity"] * orders["unit_price"]

# Suggested by an AI assistant for "average order value by region":
per_customer = orders.groupby(["region", "customer_id"])["revenue"].mean()
suggested = per_customer.groupby("region").mean()

# What the question actually asks: total revenue / number of orders
correct = (
    orders.groupby("region")["revenue"].sum()
    / orders.groupby("region")["order_id"].count()
)

print(pd.DataFrame({"suggested": suggested, "correct": correct}).round(2))
```

## Checks That Catch Most Mistakes

`sanity_checks.py`

```python
import pandas as pd

orders = pd.read_csv("../data/orders.csv")
orders["revenue"] = orders["quantity"] * orders["unit_price"]
by_region = orders.groupby("region")["revenue"].sum()

# Cheap checks that catch most wrong answers, from people or AI
assert abs(by_region.sum() - orders["revenue"].sum()) < 0.01, (
    "parts != whole"
)
assert len(by_region) == orders["region"].nunique(), "a region went missing"
assert (by_region > 0).all(), "a region has no revenue"
print("totals add up; all regions present and positive")
print(by_region.round(0).to_string())
```
